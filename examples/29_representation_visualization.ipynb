```python
from rdkit import RDLogger    
import numpy as np
from rdkit import Chem
from typing import List
from benchmark.dataset import load_dataset, LabelType
from benchmark.representations import compute_features
from benchmark.dimension_reduction import reduce_dimensions, DimensionReductionType

RDLogger.DisableLog('rdApp.*')   
```

```python
smiles, labels = load_dataset(label_types=[
  LabelType.MOL_WEIGHT,
  LabelType.QED,
  LabelType.LOGP,
])
```

```python
features = compute_features([Chem.MolFromSmiles(smile) for smile in smiles], "MorganFingerprintFeaturizer")
```

```python
pca_r = reduce_dimensions(features, DimensionReductionType.PCA)
```

```python
tsne_r = reduce_dimensions(features, DimensionReductionType.TSNE)
```

```python
umap_r = reduce_dimensions(features, DimensionReductionType.UMAP)
```

```python
plot_principal_components(pca_r, labels[LabelType.LOGP], LabelType.LOGP)
```

```python
plot_principal_components(tsne_r, labels[LabelType.LOGP], LabelType.LOGP)
```

```python
plot_principal_components(umap_r, labels[LabelType.LOGP], LabelType.LOGP)
```
